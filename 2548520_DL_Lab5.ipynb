# **RNN and LSTM for Next Word Prediction**

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, Dense
```

```python
# Load the cricket text dataset
with open("/content/cricket.txt", "r", encoding="utf-8") as file:
    text = file.read()

print("Sample Training Text:\n")
print(text[:500])
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

total_words = len(tokenizer.word_index) + 1

print("Total Vocabulary Size:", total_words)

input_sequences = []

for line in text.split('\n'):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_seq = token_list[:i+1]
        input_sequences.append(n_gram_seq)
```

```python
# Find maximum sequence length
max_seq_len = max([len(seq) for seq in input_sequences])

# Pad sequences
input_sequences = np.array(
    pad_sequences(input_sequences, maxlen=max_seq_len, padding='pre')
)

# Split predictors and label
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

# Convert labels to categorical
y = tf.keras.utils.to_categorical(y, num_classes=total_words)

print("Training Data Shape:", X.shape)
```

```python
print("\nTraining RNN Model...\n")

rnn_model.fit(X, y, epochs=50, verbose=1)
```

```python
rnn_model = Sequential()

rnn_model.add(Embedding(total_words, 64, input_length=max_seq_len-1))
rnn_model.add(SimpleRNN(128))
rnn_model.add(Dense(total_words, activation='softmax'))

rnn_model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("\nRNN Model Summary")
rnn_model.summary()

```

```python
lstm_model = Sequential()

lstm_model.add(Embedding(total_words, 64, input_length=max_seq_len-1))
lstm_model.add(LSTM(128))
lstm_model.add(Dense(total_words, activation='softmax'))

lstm_model.compile(
    loss='categorical_crossentropy',
    optimizer='adam',
    metrics=['accuracy']
)

print("\nLSTM Model Summary")
lstm_model.summary()
```

```python
print("\nTraining LSTM Model...\n")

lstm_model.fit(X, y, epochs=50, verbose=1)
```

```python
def generate_text(model, seed_text, next_words):

    for _ in range(next_words):

        token_list = tokenizer.texts_to_sequences([seed_text])[0]

        token_list = pad_sequences(
            [token_list],
            maxlen=max_seq_len-1,
            padding='pre'
        )

        predicted = np.argmax(model.predict(token_list), axis=-1)

        output_word = ""

        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break

        seed_text += " " + output_word

    return seed_text
```

```python
print("\nGenerated Text using RNN:\n")
print(generate_text(rnn_model, "cricket is", 50))
```

```python

print("\nGenerated Text using LSTM:\n")
print(generate_text(lstm_model, "cricket is", 50))
```

**Simple RNN Model:**

* Training Performance: Achieved a high accuracy of 0.9857 and a low loss of 0.0543 after 50 epochs. This indicates strong learning on the training data.
* Generated Text Quality: Produced coherent and contextually relevant text about cricket (e.g., "cricket is one of the most loved sports in many parts of the world..."). The flow of text is quite natural.

**LSTM Model:**

* Training Performance: Also performed well in training, reaching an accuracy of 0.9417 and a loss of 0.2319 after 50 epochs. Its performance was slightly lower than SimpleRNN in terms of final accuracy and loss for this specific training run.
* Generated Text Quality: The generated text exhibited significant repetition (e.g., "rate rate rate", "conditions conditions conditions"). This suggests the model might have overfitted to certain short patterns in the training data, leading to less diverse and less natural output.

**Interpretation:**

* Both models successfully learned to predict the next word to some extent, demonstrating the capability of recurrent networks for this task.
* Interestingly, for this particular dataset and model configuration, the SimpleRNN model produced more fluid and less repetitive generated text compared to the LSTM model, despite LSTMs generally being considered superior for sequence data due to their ability to capture long-term dependencies.
* The repetitive nature of the LSTM's output could be due to overfitting on the small dataset or specific hyperparameters used. Techniques like temperature sampling during generation or further hyperparameter tuning might improve the LSTM's text generation quality.
* This outcome highlights that higher training accuracy does not always directly translate to better or more creative text generation, especially with smaller datasets, and careful evaluation of generated output is crucial.
